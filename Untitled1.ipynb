```python
import glob
import os

ricerca = os.path.join(dir, "**/*.jpg")
immagini = glob.glob(ricerca, recursive=True)
immagini_pulite = [f for f in immagini if f.endswith('.jpg')]

# Creiamo il DataFrame Pandas solo con i percorsi dei file
df = pd.DataFrame(immagini_pulite, columns=["img"])

print(f"Totale immagini trovate: {len(df)}")
print(df.head())
```

```python
IMG_SIZE = 256
```

```python
from tensorflow.keras import layers, models

def build_light_landmark_model():
    model = models.Sequential([
        layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3)),

        layers.Conv2D(32,  (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(),

        layers.Conv2D(64,  (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(),

        layers.Conv2D(128, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(),

        layers.Conv2D(256, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D(),

        layers.Conv2D(256, (3,3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.GlobalAveragePooling2D(),

        layers.Dense(512, activation='relu'),
        layers.Dropout(0.4),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(7, activation='softmax')
    ])
    return model

# Creiamo il modello e stampiamo il riepilogo dei parametri
model = build_light_landmark_model()
model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(),
    metrics=['accuracy']
)
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
import os
from PIL import Image

# 1. Estraiamo le etichette dal percorso dei file
df['label'] = df['img'].apply(lambda x: x.split('/')[-2])

# Creiamo una mappatura dalle etichette testuali a numeri (ID)
unique_labels = sorted(df['label'].unique())
label_to_index = {label: idx for idx, label in enumerate(unique_labels)}
df['label_idx'] = df['label'].map(label_to_index)

print("Classi rilevate:", label_to_index)

# Filtriamo i file che non sono reali immagini valide o sono corrotti
def is_valid_image(path):
    try:
        # Controlliamo l'esistenza e la dimensione
        if not os.path.exists(path) or os.path.getsize(path) == 0:
            return False
        # Proviamo ad aprire l'immagine e a verificarne l'integrità
        with Image.open(path) as img:
            img.verify()
        return True
    except:
        return False

print(f"Totale record iniziali: {len(df)}")
df = df[df['img'].apply(is_valid_image)].reset_index(drop=True)
print(f"Totale record dopo la rimozione dei file non validi/corrotti: {len(df)}")

# 2. Funzione per caricare e preelaborare le immagini singolarmente tramite pipeline
def load_and_preprocess_image(path, label):
    image = tf.io.read_file(path)
    # Decodifichiamo in modo sicuro
    image = tf.io.decode_image(image, channels=0, expand_animations=False)

    # Gestiamo immagini in scala di grigi (1 canale) o RGBA (4 canali)
    if tf.shape(image)[-1] == 1:
        image = tf.image.grayscale_to_rgb(image)
    elif tf.shape(image)[-1] == 4:
        # Rimuoviamo il canale alpha se presente
        image = image[:, :, :3]

    # Rimuoviamo eventuali dimensioni extra di dimensione 1
    image = tf.squeeze(image)
    # Assicuriamo che l'immagine sia interpretata correttamente come un tensore 3D [H, W, C]
    image = tf.ensure_shape(image, [None, None, 3])
    image = tf.image.resize(image, [IMG_SIZE, IMG_SIZE])
    image = image / 255.0  # Normalizzazione
    return image, label

# Creiamo i dataset di TensorFlow
filepaths = df['img'].values
labels = df['label_idx'].values.astype(np.int32)

dataset = tf.data.Dataset.from_tensor_slices((filepaths, labels))
dataset = dataset.shuffle(buffer_size=len(df), reshuffle_each_iteration=False)

# Split Train / Validation (80% / 20%)
val_size = int(len(df) * 0.2)
train_dataset = dataset.skip(val_size)
val_dataset = dataset.take(val_size)

# Configurazione ottimizzata delle pipeline di input
BATCH_SIZE = 32

train_dataset = (train_dataset
                 .map(load_and_preprocess_image, num_parallel_calls=tf.data.AUTOTUNE)
                 .batch(BATCH_SIZE)
                 .prefetch(tf.data.AUTOTUNE))

val_dataset = (val_dataset
               .map(load_and_preprocess_image, num_parallel_calls=tf.data.AUTOTUNE)
               .batch(BATCH_SIZE)
               .prefetch(tf.data.AUTOTUNE))

# 3. Configurazione Early Stopping
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)

# 4. Addestramento del modello
history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=15,
    callbacks=[early_stopping]
)
```
